## HR Data Merging Summary

We loaded three datasets into something that can handle the data. These datasets are called `general_data.csv` `employee_survey_data.csv`. Manager_survey_data.csv`. We named them ` `employee` and `manager` to make it easier to work with them.

To merge the data we did two things:

1. We merged the `and `employee` datasets. We made sure that all the information from `general` was kept and we added information from `employee` where it matched. We used something called a merge to do this and we matched the data based on the `EmployeeID` column.

2. Then we merged the result with the `manager` dataset. Again we used a merge and matched the data based on the `EmployeeID` column. This added the managers information to the data.

The final merged dataset is called `merged`. It has 4410 rows and 29 columns. It has all the information about employees, surveys and managers. We noticed that some of the information is missing in columns like `NumCompaniesWorked` `TotalWorkingYears` `EnvironmentSatisfaction` `JobSatisfaction` and `WorkLifeBalance`. We might need to do something about that. We checked the `EmployeeID` column. Found that it has 4410 unique values, which means the merge was done correctly.

Finally we saved the cleaned and merged dataset, to a file called `HR_Analytics_Cleaned_Master.csv`.
