# **Executive Summary & Business Insights**
Our analysis of the 4,410 employee records gives us a clear look into our workforce makeup, pay structures, and where we might be losing people. The good news? The data is clean—no missing entries or duplicates—so we have a strong foundation to work from.

# **1. Who We Are (Demographics & Structure)**

* **A Relatively Young Workforce:** Most of our team sits in the 31–40 age bracket, with a solid entry-level presence (18–30) and fewer employees in the upper age ranges.

* **Gender Ratio:** Our headcount currently skews male (roughly 60/40 split).

* **Where People Work:** Research & Development dominates our operations, followed by Sales and HR. Naturally, our top roles reflect this—most people work as Sales Executives, Research Scientists, or Lab Technicians, with backgrounds mainly in Life Sciences and Medical fields.

* **Travel & Personal Life:** Most staff travel only occasionally, and a majority are married.

# **2. Pay & Tenure**
* **Compensation Trends:** The average monthly income sits around ~65,000 units. However, because of a small group of high earners at the top, the median is likely a better measure of standard pay.

* **Department Breakdown:** R&D and Sales show wide pay bands, while HR salaries trend slightly lower on average. Gender pay differences across departments appear minor at first glance, but are worth a deeper dive to ensure full equity.

* **Tenure:** The vast majority of employees have been with us for under 10 years, though we do have a few long-term veterans pushing up to 40 years of service.

# **3. Red Flags Around**

* **The Overall Picture:** Turnovers sit at about 16%. While that means 84% are staying put, 16% is still a significant loss of talent and budget.

**Who Is Leaving?**

* Gender doesn't seem to play a role
here—men and women leave at roughly similar rates.

* Role matters. Sales Representatives and Lab Technicians show noticeably higher turnover, pointing toward potential burnout, competitive poaching, or limited growth paths.

* Younger, lower-to-mid-income employees are the most vulnerable to leaving. Retention strategies need to focus heavily on early-career compensation and development.

# **4. How the Factors Connect**

* Unsurprisingly, Age, Total Experience, Job Level, and Pay all move together—the longer someone works and the higher up they go, the more they make.

* Outside of those traditional career-progression metrics, most other variables operate independently. This means attrition isn't driven by a single simple metric—it’s likely a mix of workplace culture, role dynamics, and compensation.
